```python

import glob
import pandas as pd

files = glob.glob("/content/*.csv")
print("CSV files found:", files)

if files:
    df = pd.read_csv(files[0])
    print("\nDataset loaded successfully!")
    print("Shape:", df.shape)
    display(df.head())
else:
    print("No CSV files found. Please upload your file using the Files panel.")
```

```python

print("Column names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nMissing values in each column:")
display(df.isnull().sum())
```

```python

print("Available categories:")
display(df["Category"].value_counts().to_frame())

print("\nAvailable subcategories:")
display(df["Subcategory"].value_counts().to_frame())

print("\nAvailable variables:")
display(df["Variable"].value_counts().to_frame())

print("\nAvailable energy sources:")
display(df["State"].dropna().unique())
```

```python

# Search for solar, wind, renewable, and electricity generation records
keywords = r"solar|wind|renewable|generation|capacity|demand"

filtered_df = df[
    df.astype(str)
      .apply(lambda col: col.str.contains(
          keywords, case=False, regex=True, na=False
      ))
      .any(axis=1)
]

display(filtered_df.head(20))

print("Matching records:", len(filtered_df))
```

```python

# Step 5: Clean the dataset

import pandas as pd

# Keep the original DataFrame unchanged
df_clean = df.copy()

# Standardize column names
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.replace(" ", "_")
)

# Remove exact duplicate rows
before = len(df_clean)
df_clean = df_clean.drop_duplicates()
after = len(df_clean)

# Convert Value to numeric
if "Value" in df_clean.columns:
    df_clean["Value"] = pd.to_numeric(
        df_clean["Value"], errors="coerce"
    )

print("Rows before cleaning:", before)
print("Rows after cleaning:", after)
print("Duplicate rows removed:", before - after)

print("\nMissing values after cleaning:")
display(df_clean.isnull().sum().to_frame("Missing_Count"))

print("\nCleaned dataset preview:")
display(df_clean.head())
```

```python

output_file = "/content/renewable_energy_analysis_cleaned.csv"

df_clean.to_csv(output_file, index=False)

print("Cleaned dataset saved successfully!")
```

```python

import matplotlib.pyplot as plt
import pandas as pd

# Convert Date into datetime format
solar_gwh["Date"] = pd.to_datetime(solar_gwh["Date"])

# Select India's overall total, if available
print(solar_gwh["State_type"].unique())
```

```python

india_solar = solar_gwh[
    solar_gwh["State_type"].str.lower() == "total"
].copy()

india_solar = india_solar.sort_values("Date")

plt.figure(figsize=(12, 5))
plt.plot(
    india_solar["Date"],
    india_solar["Value"],
    marker="o"
)

plt.title("India's Monthly Solar Electricity Generation")
plt.xlabel("Date")
plt.ylabel("Solar Generation (GWh)")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python

wind = df_clean[
    (df_clean["Category"] == "Electricity generation") &
    (df_clean["Variable"] == "Wind") &
    (df_clean["Unit"] == "GWh") &
    (df_clean["State_type"].str.lower() == "total")
].copy()

wind["Date"] = pd.to_datetime(wind["Date"])

display(wind.head(5))

print("Total wind records:", len(wind))
```

```python

wind = wind.sort_values("Date")

plt.figure(figsize=(12, 5))
plt.plot(
    wind["Date"],
    wind["Value"],
    marker="o"
)

plt.title("India's Monthly Wind Electricity Generation")
plt.xlabel("Date")
plt.ylabel("Wind Generation (GWh)")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(12, 5))

plt.plot(
    india_solar["Date"],
    india_solar["Value"],
    label="Solar",
    marker="o"
)

plt.plot(
    wind["Date"],
    wind["Value"],
    label="Wind",
    marker="o"
)

plt.title("India: Monthly Solar vs Wind Generation")
plt.xlabel("Date")
plt.ylabel("Electricity Generation (GWh)")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python

print("Total Solar Generation:", round(india_solar["Value"].sum(), 2), "GWh")
print("Total Wind Generation:", round(wind["Value"].sum(), 2), "GWh")

print("Average Monthly Solar Generation:",
      round(india_solar["Value"].mean(), 2), "GWh")

print("Average Monthly Wind Generation:",
      round(wind["Value"].mean(), 2), "GWh")
```

```python

total_solar = india_solar["Value"].sum()
total_wind = wind["Value"].sum()

combined_total = total_solar + total_wind

print("Solar share:", round(total_solar / combined_total * 100, 2), "%")
print("Wind share:", round(total_wind / combined_total * 100, 2), "%")
```

```python

plt.figure(figsize=(7, 7))

plt.pie(
    [total_solar, total_wind],
    labels=["Solar", "Wind"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Solar vs Wind Share of Combined Generation")
plt.axis("equal")
plt.show()
```

```python

india_solar["Year"] = india_solar["Date"].dt.year

yearly_solar = (
    india_solar.groupby("Year")["Value"]
    .sum()
    .reset_index()
)

yearly_solar.columns = ["Year", "Solar_Generation_GWh"]

display(yearly_solar)
```

```python

wind["Year"] = wind["Date"].dt.year

yearly_wind = (
    wind.groupby("Year")["Value"]
    .sum()
    .reset_index()
)

yearly_wind.columns = ["Year", "Wind_Generation_GWh"]

display(yearly_wind)
```

```python

yearly_comparison = pd.merge(
    yearly_solar,
    yearly_wind,
    on="Year",
    how="inner"
)

plt.figure(figsize=(10, 5))

plt.plot(
    yearly_comparison["Year"],
    yearly_comparison["Solar_Generation_GWh"],
    marker="o",
    label="Solar"
)

plt.plot(
    yearly_comparison["Year"],
    yearly_comparison["Wind_Generation_GWh"],
    marker="o",
    label="Wind"
)

plt.title("India: Yearly Solar vs Wind Generation")
plt.xlabel("Year")
plt.ylabel("Generation (GWh)")
plt.xticks(yearly_comparison["Year"])
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python

yearly_comparison = yearly_comparison.sort_values("Year").copy()

yearly_comparison["Solar_Growth_%"] = (
    yearly_comparison["Solar_Generation_GWh"].pct_change() * 100
)

yearly_comparison["Wind_Growth_%"] = (
    yearly_comparison["Wind_Generation_GWh"].pct_change() * 100
)

display(
    yearly_comparison[
        ["Year", "Solar_Growth_%", "Wind_Growth_%"]
    ].round(2)
)
```

```python

growth_data = yearly_comparison.dropna(
    subset=["Solar_Growth_%", "Wind_Growth_%"]
)

plt.figure(figsize=(10, 5))

plt.plot(
    growth_data["Year"],
    growth_data["Solar_Growth_%"],
    marker="o",
    label="Solar Growth"
)

plt.plot(
    growth_data["Year"],
    growth_data["Wind_Growth_%"],
    marker="o",
    label="Wind Growth"
)

plt.axhline(0, linewidth=1)
plt.title("Year-wise Solar vs Wind Generation Growth")
plt.xlabel("Year")
plt.ylabel("Growth (%)")
plt.xticks(growth_data["Year"])
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python

print("Solar latest date:", india_solar["Date"].max())
print("Wind latest date:", wind["Date"].max())

print("\nSolar records per year:")
display(india_solar.groupby("Year")["Date"].nunique())

print("\nWind records per year:")
display(wind.groupby("Year")["Date"].nunique())
```

```python

india_solar["Year"] = india_solar["Date"].dt.year
wind["Year"] = wind["Date"].dt.year

solar_compare = india_solar[
    (india_solar["Year"].isin([2024, 2025])) &
    (india_solar["Date"].dt.month <= 11)
].groupby("Year")["Value"].sum()

wind_compare = wind[
    (wind["Year"].isin([2024, 2025])) &
    (wind["Date"].dt.month <= 11)
].groupby("Year")["Value"].sum()

comparison = pd.DataFrame({
    "Solar_Generation_GWh": solar_compare,
    "Wind_Generation_GWh": wind_compare
})

display(comparison.round(2))
```

```python

solar_growth = (
    (comparison.loc[2025, "Solar_Generation_GWh"] /
     comparison.loc[2024, "Solar_Generation_GWh"]) - 1
) * 100

wind_growth = (
    (comparison.loc[2025, "Wind_Generation_GWh"] /
     comparison.loc[2024, "Wind_Generation_GWh"]) - 1
) * 100

print(f"Solar growth (Jan-Nov): {solar_growth:.2f}%")
print(f"Wind growth (Jan-Nov): {wind_growth:.2f}%")
```

```python

sources = ["Solar", "Wind"]
growth_rates = [solar_growth, wind_growth]

plt.figure(figsize=(7, 5))
bars = plt.bar(sources, growth_rates)

plt.title("Solar vs Wind Growth: Jan–Nov 2024 vs Jan–Nov 2025")
plt.ylabel("Generation Growth (%)")

for bar, rate in zip(bars, growth_rates):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{rate:.2f}%",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

```python

state_solar = df_clean[
    (df_clean["Category"] == "Electricity generation") &
    (df_clean["Variable"] == "Solar") &
    (df_clean["Unit"] == "GWh") &
    (df_clean["State_type"].str.lower() == "state") &
    (df_clean["Date"].dt.year == 2024)
].copy()

print("Number of records:", len(state_solar))
display(state_solar.head(10))
```

```python

# Filter 2024 state-level wind generation data
state_wind = df_clean[
    (df_clean["Category"] == "Electricity generation") &
    (df_clean["Variable"] == "Wind") &
    (df_clean["Unit"] == "GWh") &
    (df_clean["State_type"].str.lower() == "state") &
    (df_clean["Date"].dt.year == 2024)
].copy()

# Find the top 10 wind-generating states
top10_wind = (
    state_wind.groupby("State")["Value"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

top10_wind.columns = ["State", "Wind_Generation_GWh"]

display(top10_wind)
```

```python

# Check available installed capacity data

capacity_data = df_clean[
    df_clean["Category"] == "Capacity"
].copy()

print("Total capacity records:", len(capacity_data))

print("\nAvailable energy variables:")
print(sorted(capacity_data["Variable"].dropna().unique()))

print("\nAvailable units:")
print(sorted(capacity_data["Unit"].dropna().unique()))

print("\nSample records:")
display(capacity_data.head(10))
```

```python

# Filter national-level installed capacity data
india_capacity = df_clean[
    (df_clean["Category"] == "Capacity") &
    (df_clean["State_type"].str.lower() == "total") &
    (df_clean["Unit"] == "MW") &
    (df_clean["Variable"].isin([
        "Solar", "Wind", "Hydro", "Bioenergy",
        "Nuclear", "Coal", "Gas", "Other Fossil",
        "Other Renewables", "Renewables", "Fossil", "Clean"
    ]))
].copy()

print("National capacity records:", len(india_capacity))

print("\nAvailable national capacity dates:")
print(india_capacity["Date"].min(), "to", india_capacity["Date"].max())

print("\nLatest records:")
display(
    india_capacity[
        india_capacity["Date"] == india_capacity["Date"].max()
    ][["Date", "Variable", "Unit", "Value"]]
)
```

```python

import matplotlib.pyplot as plt

# Latest national capacity date
latest_date = india_capacity["Date"].max()

# Select individual energy sources
capacity_chart = india_capacity[
    (india_capacity["Date"] == latest_date) &
    (india_capacity["Variable"].isin([
        "Solar", "Wind", "Hydro", "Bioenergy", "Nuclear"
    ]))
].dropna(subset=["Value"]).copy()

# Sort by capacity
capacity_chart = capacity_chart.sort_values(
    "Value", ascending=True
)

plt.figure(figsize=(10, 6))
plt.barh(capacity_chart["Variable"], capacity_chart["Value"])

plt.title("India: Installed Capacity by Energy Source")
plt.xlabel("Installed Capacity (MW)")
plt.ylabel("Energy Source")
plt.tight_layout()
plt.show()
```

```python

# Solar and Wind installed capacity over time
capacity_trend = india_capacity[
    (india_capacity["Variable"].isin(["Solar", "Wind"])) &
    (india_capacity["Value"].notna())
].copy()

# Keep one record per date and energy source
capacity_trend = capacity_trend.sort_values("Date")

# Show starting and latest capacity
for source in ["Solar", "Wind"]:
    source_data = capacity_trend[
        capacity_trend["Variable"] == source
    ].sort_values("Date")

    first = source_data.iloc[0]
    latest = source_data.iloc[-1]

    growth_pct = (
        (latest["Value"] - first["Value"]) / first["Value"]
    ) * 100

    print(f"\n{source} Installed Capacity")
    print(f"Starting date: {first['Date'].date()}")
    print(f"Starting capacity: {first['Value']:,.2f} MW")
    print(f"Latest date: {latest['Date'].date()}")
    print(f"Latest capacity: {latest['Value']:,.2f} MW")
    print(f"Total growth: {growth_pct:.2f}%")
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

for source in ["Solar", "Wind"]:
    source_data = capacity_trend[
        capacity_trend["Variable"] == source
    ].sort_values("Date")

    plt.plot(
        source_data["Date"],
        source_data["Value"],
        label=source,
        linewidth=2
    )

plt.title("India's Solar vs Wind Installed Capacity (2019–2025)")
plt.xlabel("Date")
plt.ylabel("Installed Capacity (MW)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

# Check national power-sector emissions data
emissions_data = df_clean[
    (df_clean["Category"] == "Power sector emissions") &
    (df_clean["State_type"].str.lower() == "total")
].copy()

print("Total national emissions records:", len(emissions_data))

print("\nAvailable emission variables:")
print(sorted(emissions_data["Variable"].dropna().unique()))

print("\nAvailable units:")
print(sorted(emissions_data["Unit"].dropna().unique()))

print("\nSample emissions records:")
display(emissions_data.head(10))
```

```python

# Filter total power-sector CO2 emissions
total_emissions = emissions_data[
    (emissions_data["Variable"] == "Total emissions") &
    (emissions_data["Unit"] == "ktCO2")
].copy()

total_emissions = total_emissions.dropna(
    subset=["Date", "Value"]
).sort_values("Date")

print("First available record:")
display(total_emissions.head(1)[["Date", "Variable", "Value", "Unit"]])

print("\nLatest available record:")
display(total_emissions.tail(1)[["Date", "Variable", "Value", "Unit"]])

# Plot emissions trend
plt.figure(figsize=(12, 6))

plt.plot(
    total_emissions["Date"],
    total_emissions["Value"],
    linewidth=2
)

plt.title("India Power Sector: Total CO2 Emissions (2019–2025)")
plt.xlabel("Date")
plt.ylabel("Total Emissions (ktCO2)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

# Filter national CO2 intensity data
co2_intensity = emissions_data[
    (emissions_data["Variable"] == "CO2 intensity") &
    (emissions_data["Unit"] == "gCO2/kWh")
].copy()

co2_intensity = co2_intensity.dropna(
    subset=["Date", "Value"]
).sort_values("Date")

print("First available CO2 intensity record:")
display(
    co2_intensity.head(1)[["Date", "Variable", "Value", "Unit"]]
)

print("\nLatest available CO2 intensity record:")
display(
    co2_intensity.tail(1)[["Date", "Variable", "Value", "Unit"]]
)

# Plot CO2 intensity trend
plt.figure(figsize=(12, 6))

plt.plot(
    co2_intensity["Date"],
    co2_intensity["Value"],
    linewidth=2
)

plt.title("India Power Sector: CO2 Intensity (2019–2025)")
plt.xlabel("Date")
plt.ylabel("CO2 Intensity (gCO2/kWh)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
